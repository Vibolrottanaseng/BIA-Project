```python
import re
import math
from urllib.parse import urlparse, parse_qs
import pandas as pd

def shannon_entropy(text):
    if not text:
        return 0.0
    probs = [text.count(char) / len(text) for char in set(text)]
    return -sum(p * math.log2(p) for p in probs)

def extract_url_features(url):
    url = str(url)
    parsed = urlparse(url)

    domain = parsed.netloc
    path = parsed.path
    query = parsed.query

    # remove port
    domain_no_port = domain.split(":")[0]

    # split domain parts
    domain_parts = domain_no_port.split(".") if domain_no_port else []

    return {
        "url_length": len(url),
        "has_ip_address": int(bool(re.fullmatch(r"(\\d{1,3}\\.){3}\\d{1,3}", domain_no_port))),
        "dot_count": url.count("."),
        "https_flag": int(parsed.scheme.lower() == "https"),
        "url_entropy": shannon_entropy(url),
        "token_count": len([x for x in re.split(r"[/:?&=._-]", url) if x]),
        "subdomain_count": max(len(domain_parts) - 2, 0) if len(domain_parts) >= 2 else 0,
        "query_param_count": len(parse_qs(query)),
        "tld_length": len(domain_parts[-1]) if len(domain_parts) >= 1 else 0,
        "path_length": len(path),
        "has_hyphen_in_domain": int("-" in domain_no_port),
        "number_of_digits": sum(c.isdigit() for c in url),
        "suspicious_file_extension": int(path.lower().endswith((".exe", ".zip", ".rar", ".scr", ".js", ".bat"))),
        "domain_name_length": len(domain_no_port),
        "percentage_numeric_chars": sum(c.isdigit() for c in url) / len(url) if len(url) > 0 else 0.0,
    }

def build_feature_df(df, url_col="URL", label_col="label"):
    df = df.copy()

    # force numeric labels; invalid values become NaN
    df[label_col] = pd.to_numeric(df[label_col], errors="coerce")

    # drop rows with missing labels or missing URLs
    df = df.dropna(subset=[url_col, label_col]).copy()

    # convert labels to int
    df[label_col] = df[label_col].astype(int)

    # extract features
    X = pd.DataFrame(df[url_col].apply(extract_url_features).tolist()).reset_index(drop=True)
    y = df[label_col].reset_index(drop=True)

    return X, y
```

```python
print("Train label values:")
print(train_df["label"].value_counts(dropna=False))

print("\nTest label values:")
print(test_df["label"].value_counts(dropna=False))
```

```python
train_df["label"] = pd.to_numeric(train_df["label"], errors="coerce")
test_df["label"] = pd.to_numeric(test_df["label"], errors="coerce")

print("Missing labels in train:", train_df["label"].isna().sum())
print("Missing labels in test:", test_df["label"].isna().sum())

train_df = train_df.dropna(subset=["label"]).copy()
test_df = test_df.dropna(subset=["label"]).copy()

train_df["label"] = train_df["label"].astype(int)
test_df["label"] = test_df["label"].astype(int)

print("\nCleaned train labels:")
print(train_df["label"].value_counts(dropna=False))

print("\nCleaned test labels:")
print(test_df["label"].value_counts(dropna=False))
```

```python
X_train, y_train = build_feature_df(train_df, url_col="URL", label_col="label")
X_test, y_test = build_feature_df(test_df, url_col="URL", label_col="label")

print("Train features shape:", X_train.shape)
print("Test features shape:", X_test.shape)
print(X_train.head())
```

```python
train_features = pd.concat([X_train, y_train.rename("label")], axis=1)

class_0 = train_features[train_features["label"] == 0]
class_1 = train_features[train_features["label"] == 1]

min_size = min(len(class_0), len(class_1))

class_0_balanced = class_0.sample(n=min_size, random_state=42)
class_1_balanced = class_1.sample(n=min_size, random_state=42)

train_balanced = pd.concat([class_0_balanced, class_1_balanced], axis=0)
train_balanced = train_balanced.sample(frac=1, random_state=42).reset_index(drop=True)

X_train_balanced = train_balanced.drop(columns=["label"])
y_train_balanced = train_balanced["label"]

print("Balanced train labels:")
print(y_train_balanced.value_counts())
```

```python
# Train on dataset 1 test on dataset 2

from sklearn.ensemble import RandomForestClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, roc_auc_score, classification_report

# Random Forest
rf = RandomForestClassifier(n_estimators=100, random_state=42)
rf.fit(X_train_balanced, y_train_balanced)

y_pred_rf = rf.predict(X_test)
y_prob_rf = rf.predict_proba(X_test)[:, 1]

print("Random Forest External Test")
print("Accuracy :", accuracy_score(y_test, y_pred_rf))
print("Precision:", precision_score(y_test, y_pred_rf))
print("Recall   :", recall_score(y_test, y_pred_rf))
print("F1-score :", f1_score(y_test, y_pred_rf))
print("ROC-AUC  :", roc_auc_score(y_test, y_prob_rf))
print("\nClassification Report:\n", classification_report(y_test, y_pred_rf))

# Logistic Regression
lr = LogisticRegression(max_iter=1000, random_state=42)
lr.fit(X_train_balanced, y_train_balanced)

y_pred_lr = lr.predict(X_test)
y_prob_lr = lr.predict_proba(X_test)[:, 1]

print("\nLogistic Regression External Test")
print("Accuracy :", accuracy_score(y_test, y_pred_lr))
print("Precision:", precision_score(y_test, y_pred_lr))
print("Recall   :", recall_score(y_test, y_pred_lr))
print("F1-score :", f1_score(y_test, y_pred_lr))
print("ROC-AUC  :", roc_auc_score(y_test, y_prob_lr))
print("\nClassification Report:\n", classification_report(y_test, y_pred_lr))
```
